```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/kaggle/input/datasets/devbhise/quote-of-genius/qoute_dataset.csv")
```

```python
quotes = df["quote"]
```

```python
quotes.head()
```

```python
quotes = quotes.str.lower()
```

```python
import string
translator = str.maketrans('','',string.punctuation)
quotes = quotes.apply(lambda x:x.translate(translator))
```

```python
quotes.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size = 8978

tokenizer = Tokenizer(num_words = vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index = tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequence = tokenizer.texts_to_sequences(quotes)
```

```python
quotes[0]
```

```python
sequence[0]
```

```python
for i in range(3):
    print(quotes[i])
```

```python
for i in range(4):
    print(sequence[i])
```

```python
X = []
y = []

for seq in sequence:
    for i in range(1,len(seq)):
        input_seq = seq[:i]
        output_seq = seq[i]
        X.append(input_seq)
        y.append(output_seq)
        
```

```python
X
```

```python
len(X)
```

```python
len(y)
```

```python
max_len = max(len(x) for x in X)
print(max_len)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded = pad_sequences(X,maxlen=max_len,padding = 'pre')
```

```python
X_padded[1]
```

```python
y = np.array(y)
```

```python
X_padded.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y,num_classes=vocab_size)
```

```python
y_one_hot.shape
```

```python
# Simple RNN

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense,SimpleRNN
```

```python
embeddings_dim = 50
rnn_units = 128
```

```python
rnn_model = Sequential()
rnn_model.add(
    Embedding(input_dim=vocab_size,output_dim=embeddings_dim,input_length=max_len)
)
rnn_model.add(
    SimpleRNN(units = rnn_units)
)
rnn_model.add(Dense(units=vocab_size,activation='softmax'))
```

```python
rnn_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
rnn_model.summary()
```

```python
lstm_model = Sequential()
lstm_model.add(
    Embedding(input_dim=vocab_size,output_dim=embeddings_dim,
             input_length=max_len)
)
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=vocab_size,activation='softmax'))
```

```python
lstm_model.compile(
    optimizer = 'adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
lstm_model.summary()
```

```python
epochs = 100
batch_size = 128
```

```python
history_lstm = lstm_model.fit(
     X_padded,y_one_hot,
    epochs=epochs,
    batch_size = batch_size,
    validation_split=0.1
)
```

```python
history_rnn = rnn_model.fit(
    X_padded,y_one_hot,
    epochs=epochs,
    batch_size = batch_size,
    validation_split=0.1
)
```
